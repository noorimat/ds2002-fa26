### How to read an error before you fix it

Python's traceback is read **bottom up**. The last line names the exception and the problem; the line above it points at your code.

```
Traceback (most recent call last):
  File "<stdin>", line 2, in <module>
    print(greeting)
NameError: name 'greeting' is not defined
```

Four exceptions cover most of what you will hit this semester:

| Exception | Usually means |
|---|---|
| `NameError` | The variable was never created, or the cell that created it never ran |
| `TypeError` | You mixed types — text where a number belongs is the classic |
| `KeyError` | That column or dictionary key does not exist. Check spelling and case |
| `FileNotFoundError` | The path is wrong for *this* environment |

### Bug 2 — text where a number belongs

This should print `30`. Note that `'10' + 20` and `'10' + '20'` fail in different ways — make sure you know which one you are looking at.

````python
total = int('10') + 20
print(total)
````

### Bug 6 — the answer changes when you re-run it

Run the next cell three times and watch the total move. Then fix it using what you built Wednesday: leave the source data alone and derive a new column from it. Your fix must give the same total on every run.

````python
sales = pd.DataFrame({
    'game': ['NC State', 'Louisville'],
    'units': [1450, 1610],
})

work = sales.copy()
work['units_with_fee'] = work['units'] * 1.05
print('total units:', round(work['units_with_fee'].sum(), 2))
````

````python
# Proof — this must print the same number three times
for _ in range(3):
    work = sales.copy()
    work['units_with_fee'] = work['units'] * 1.05
    print('total units:', round(work['units_with_fee'].sum(), 2))
````

**What was wrong:** the original code overwrote `sales['units']` in place, so every re-run multiplied it again by 1.05, deriving a new column from an untouched source fixes it.

### Stretch

Write a `check_orders(df)` function that raises a clear error if the frame is missing either expected column, has zero rows, or contains a negative value in `units`. Then call it on `orders`. This is the shape of validation you will want in both projects.

````python
# stretch
def check_orders(df):
    required = {'game', 'units'}
    missing = required - set(df.columns)
    if missing:
        raise ValueError(f'missing columns: {missing}')
    if len(df) == 0:
        raise ValueError('orders is empty')
    if (df['units'] < 0).any():
        raise ValueError('units contains a negative value')

check_orders(orders)
print('orders passed validation.')
````
